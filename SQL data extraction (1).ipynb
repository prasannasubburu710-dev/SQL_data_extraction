```python
import sqlite3
import pandas as pd
```

```python
# Create a connection to SQLite database
conn = sqlite3.connect("sales_data.db")

print("Database created successfully!")
```

```python
# Create a sample sales dataset

data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile',
                'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop'],
    
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics',
                 'Electronics', 'Electronics', 'Electronics', 'Electronics',
                 'Electronics', 'Electronics'],
    
    'Customer': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
                 'Frank', 'Grace', 'Henry', 'Ivy', 'Jack'],
    
    'Sales': [50000, 30000, 20000, 55000, 35000,
              22000, 60000, 40000, 25000, 65000],
    
    'Month': ['January', 'January', 'January', 'February', 'February',
              'February', 'March', 'March', 'March', 'April']
}

df = pd.DataFrame(data)

df
```

```python
# Import the DataFrame into SQLite

df.to_sql('sales', conn, if_exists='replace', index=False)

print("Dataset imported successfully into SQLite!")
```

```python
query = """
SELECT sales.Product, sales.Customer, sales.Sales, customers.City
FROM sales
JOIN customers
ON sales.Customer = customers.Customer
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT Product, Customer, Sales
FROM sales
WHERE Sales > (
    SELECT AVG(Sales)
    FROM sales
)
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
WITH ProductSales AS (
    SELECT Product, SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Product
)
SELECT *
FROM ProductSales
WHERE Total_Sales > 100000
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT Product, Customer, Sales,
       RANK() OVER (ORDER BY Sales DESC) AS Sales_Rank
FROM sales
"""

result = pd.read_sql(query, conn)

result
```

```python
conn.execute("DROP VIEW IF EXISTS High_Sales")

query = """
CREATE VIEW High_Sales AS
SELECT Product, Customer, Sales
FROM sales
WHERE Sales > 40000
"""

conn.execute(query)

print("View created successfully!")
```

```python
query = "SELECT * FROM High_Sales"

result = pd.read_sql(query, conn)

result
```
